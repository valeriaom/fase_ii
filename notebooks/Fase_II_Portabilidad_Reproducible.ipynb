## Interpretación de la matriz de confusión

Los falsos negativos corresponden a salidas sintéticas no detectadas y los falsos positivos a alertas sin salida observada. Esta interpretación no puede extrapolarse a clientes reales.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay
cm = np.asarray(metadata['test_metrics']['confusion_matrix'], dtype=int)
fig, ax = plt.subplots(figsize=(7, 5))
ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Sin salida','Salida']).plot(cmap='Blues', values_format='d', ax=ax)
ax.set_title('Evaluación sobre datos sintéticos')
plt.tight_layout()
plt.show()
```
